```python
#String slicing02
x="Python programming"
print(x[1:])
print(x[0:])
print(x[:6])
print(x[-1])
print(x[-1:])
```

```python
#String slicing03
x="Python programming"
print(x[-6:])
print(x[-6:2:-2])
print(x[-6::-2])
print(x[2::8])
```

```python
#String method
s="Artificial Intelligence and Machine Learning"
print(s.upper())
print(s.lower())
print(s.find("Machine"))
print(s.replace("Intelligence","Foundation"))
print(len(s))
```

```python
# User-Defined Functions for String Operation
def string_length(s):
    return len(s)
def to_upper(s):
    return s.upper()
def to_lower(s):
    return s.lower()
def reverse_string(s):
    return s[::-1]
text = input("Enter a string:")
print("Length of the string:", string_length(text))
print("Upper case:",to_upper(text))
print("Lower case:",to_lower(text))
print("Reversed String:",reverse_string(text))
```
