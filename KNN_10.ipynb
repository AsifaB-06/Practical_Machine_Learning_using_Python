```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import matplotlib.pyplot as plt

# Input data
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create a KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(x, y)

# Predict the result for x = 6
predict_m = model.predict([[6]])

# Display the results
print("Predicted result:", predict_m)

# Plot the data
plt.scatter(x[y == 0], y[y == 0], label="Fail")
plt.scatter(x[y == 1], y[y == 1], label="Pass")

# Plot the predicted point
plt.scatter(6, predict_m[0], marker="*", s=200, label="Prediction")

plt.xlabel("Study Hours")
plt.ylabel("Result")
plt.title("KNN Classification - Study Hours vs Result")
plt.legend()
plt.show()
```
