```python
X = [[1,2],[2,3],[3,4],[4,5]]
Y = [0,0,1,1]
```

```python
# INITIALIZE THE WEIGHTS
w1 = 0.5
w2 = 0.5
b = -2
```

```python
import math
learning_rate = 0.1
def sigmoid(n):
    return 1/1+math.exp(-n)
```

```python
for epoch in range(1000):
    for i in range(len(X)):
        x1=X[i][0]
        x2=X[i][1]
        #FORWARD PROPAGATION
        z=x1*w1+x2*w2+b
        #PREDICTED OUTPUT
        output=sigmoid(z)
        #ERROR
        error=Y[i]-output
        #BACK PROPAGATION
        gradient=error*output*(1-output)
        #UPDATE WEIGHTS
        w1=w1+learning_rate*gradient*x1
        w2=w2+learning_rate*gradient*x2
        b=b+learning_rate*gradient
```

```python
#DISPLAY OUTPUT
for i in range(len(X)):
    z=X[i][0]*w1+X[i][1]*w2+b
    output = sigmoid(z)
    print(X[i],"Actual=",Y[i],"Predicted=",round(output,2))
```
